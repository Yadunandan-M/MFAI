let $h$ = heights and $w$ = weights

```python
h = [1.7, 1.62, 1.52, 1.85, 1.91, 1.42]
w = [72, 64, 84, 80, 72, 70]
```

find_mean() takes in a list of data and returns the mean.

```python
def find_mean(list_of_values):
    sum_of_values = 0
    for value in list_of_values:
        sum_of_values += value
    return sum_of_values / len(list_of_values)
```

find_differences() takes in a list of values and its mean and returns a list of differences of the datapoints from the mean.

```python
def find_differences(list_of_values, mean):
    differences = []
    for value in list_of_values:
        differences.append(value - mean)
    return differences
```

find_differences_squared() takes in a list of values and returns a list of squared values of the datapoints

```python
def find_differences_squared(list_of_values):
    differences_squared = []
    for value in list_of_values:
        differences_squared.append(value**2)
    return differences_squared
```

find_XY() takes in 2 lists and returns the list where member at an index is a product of members at same index of the input lists.

```python
def find_XY(X_list, Y_list):
    XY_list = []
    if len(X_list) == len(Y_list):
        for index in range(len(X_list)):
            XY_list.append(X_list[index] * Y_list[index])
        return XY_list
    else:
        print("error: lengths of lists dont match")
```

find_sum() returns sum of elements of a list

```python
def find_sum(list_of_values):
    _sum = 0
    for value in list_of_values:
        _sum+=value
    return _sum
```

finding mean of heights $\overline{h}$

```python
mean_height = find_mean(heights)
print(mean_height)
```

finding mean of weights $\overline{w}$

```python
mean_weight  = find_mean(weights)
print(mean_weight)
```

let $X = h_i - \overline{h}$ and $Y = w_i - \overline{w}$

```python
X = find_differences(h, mean_height)
print("X = ", X)
Y = find_differences(w, mean_weight)
print("Y = ", Y)
```

finding $X^2$ and $Y^2$

```python
X_squared = find_differences_squared(X)
print("X_SQUARED =", X_squared)
Y_squared = find_differences_squared(Y)
print("Y_SQUARED =", Y_squared)
```

finding $XY$

```python
XY = find_XY(X, Y)
print("XY =", XY)
```

covariance matrix = $\begin{bmatrix} \frac{\sum X^2}{n-1} &  \frac{\sum XY}{n-1} \\  \frac{\sum XY}{n-1} & \frac{\sum Y^2}{n-1}\end{bmatrix}$

where $n$ = number of datapoints

```python
n = len(h)
i_11 = find_sum(X_squared)/(n-1)
i_22 = find_sum(Y_squared)/(n-1)
i_12 = find_sum(XY)/(n-1)
covariance_matrix = [[i_11, i_12], [i_12, i_22]]
print("covariance matrix = ", covariance_matrix)
```
