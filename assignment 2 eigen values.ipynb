Data for 3 variables

```python
X = [12.1, 13.2, 15.6, 17.2, 18.8, 10.3, 11.7, 16.4]
Y = [48, 59, 32, 18, 41, 32, 31, 30]
Z = [101, 171, 112, 132, 140, 112, 151, 96]
```

```python
def calculate_mean(array):
    sum = 0
    for element in array:
        sum += element
    return sum/len(array)
```

```python
mean_X = calculate_mean(X)
mean_Y = calculate_mean(Y)
mean_Z = calculate_mean(Z)
print("The mean of X is: ", mean_X)
print("The mean of Y is: ", mean_Y)
print("The mean of Z is: ", mean_Z)
```

```python
def calculate_variance(array):
    mean = calculate_mean(array)
    sum = 0
    for element in array:
        deviation = element - mean
        sum += deviation **2
    variance = sum / (len(array) - 1)
    return variance
```

```python
def calculate_covariance(array1, array2):
    mean1 = calculate_mean(array1)
    mean2 = calculate_mean(array2)

    sum = 0
    for index in range(len(array1)):
        variable1 = array1[index]
        variable2 = array2[index]
        deviation1 = variable1 - mean1
        deviation2 = variable2 - mean2

        sum += deviation1 * deviation2
    covariance = sum / (len(array1) - 1)
    return covariance
```

```python
X_variance = calculate_variance(X)
Y_variance = calculate_variance(Y)
Z_variance = calculate_variance(Z)
XY_covariance = calculate_covariance(X, Y)
YZ_covariance = calculate_covariance(Y, Z)
ZX_covariance = calculate_covariance(Z, X)
print("variance in X: ", X_variance)
print("variance in Y: ", Y_variance)
print("variance in Z: ", Z_variance)
print("covariance between X and Y: ", XY_covariance)
print("covariance between Y and Z: ", YZ_covariance)
print("covariance between Z and X: ", ZX_covariance)
```

```python
covariance_matrix = [[X_variance, XY_covariance, ZX_covariance],
                     [XY_covariance, Y_variance, YZ_covariance],
                     [ZX_covariance, YZ_covariance, Z_variance]]
print("the covariance matrix is:\n", covariance_matrix)
```

```python
import numpy as np
cov_matrix = np.array(covariance_matrix)
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)
print("eigen values are ", eigenvalues)
print("eigen vectors are ", eigenvectors)
```

```python
cov_matrix_inverse = np.linalg.inv(cov_matrix)
print("the inverse of covariance matrix is ", cov_matrix_inverse)
```
